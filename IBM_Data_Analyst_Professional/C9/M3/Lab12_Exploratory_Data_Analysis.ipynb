#### Step 1: Install and Import Libraries


Install the necessary libraries for data manipulation and visualization.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
## First let's view these missing values:
cols = ["Employment", "JobSat", "RemoteWork"]
print(df[cols].isnull().sum())
```

```python
print(df["JobSat"].value_counts())
```

```python
print(df["RemoteWork"].value_counts())
```

```python
# Consider JobSat is a job satisfaction, its value range from 0-10, we can replace missing values as its mean value
mean_jobsat = df["JobSat"].mean()
print(mean_jobsat)

# Later in EDA, it's clear that there are too much null values to estimate as mean, instead, we drop these columns
df = df.dropna(subset=["JobSat"], axis=0)
```

```python
# RemoteWork is a categorical statistic, therefore we replace its null values with the mode
mode_remotework = df["RemoteWork"].value_counts().idxmax()
print(mode_remotework)

df["RemoteWork"] = df["RemoteWork"].fillna(mode_remotework)
df["RemoteWork"].isnull().sum()
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
## Write your code here
df["YearsCodePro"] = pd.to_numeric(df["YearsCodePro"], errors="coerce")
df["YearsCodePro"] = df["YearsCodePro"].fillna(0)

df["YearsCodePro_Binned"] = pd.cut(
    df["YearsCodePro"], 
    bins=[0, 5, 10, 20, 100], 
    labels=["0-5 Years", "5-10 Years", "10-20 Years", ">20 Years"],
    include_lowest=True
)

df[["YearsCodePro", "YearsCodePro_Binned"]].head()
```

```python
median_jobsat = (
    df.groupby("YearsCodePro_Binned", observed=False)["JobSat"]
    .median()
    .reset_index()
)

median_jobsat
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    median_jobsat["YearsCodePro_Binned"].astype(str),
    median_jobsat["JobSat"]
)

plt.xlabel("Years of Professional Coding Experience")
plt.ylabel("Median Job Satisfaction")
plt.title("Median Job Satisfaction By Professional Coding Experience")
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
# Check countries
print(df["Country"].value_counts().to_string())

# 166 countries would be too overwhelming, instead we select 3 countries
countries = ["Canada", "United States of America", "United Kingdom of Great Britain and Northern Ireland", "Germany"]

df_regions = df[df["Country"].isin(countries)].copy()
df_regions = df_regions[["Country", "LanguageHaveWorkedWith"]]
```

```python
# Check for values of "LanguageHaveWorkedWith":
print(df["LanguageHaveWorkedWith"].value_counts())

# As seen the format are unordered lists separated by ";", we need to split the programming languages
languages = (
    df_regions
    .assign(Language=df_regions["LanguageHaveWorkedWith"].str.split(";"))
    .explode("Language")
    .reset_index(drop=True)
)

languages["Language"] = languages["Language"].str.strip()
languages
```

```python
language_counts = pd.crosstab(
    languages["Language"],
    languages["Country"]
)

print(language_counts.shape)
language_counts
```

```python
# 49 different languages is too much for a heatmap, instead we narrow it down to the top 10
top_10_languages = languages["Language"].value_counts().head(10).index
print(top_10_languages)

heatmap_data = pd.crosstab(
    languages["Language"],
    languages["Country"]
).loc[top_10_languages]

heatmap_data
```

```python
plt.figure(figsize=(8, 5))
sns.heatmap(data=heatmap_data)
```

```python
# Seen from the last code block, we need to format the country label, the names are too long
country_labels = ["Canada", "Germany", "UK & Ireland", "USA"]

plt.figure(figsize=(8, 5))

# Also, add values and change the cmap to a monochrome for easy reading
sns.heatmap(
    data=heatmap_data, 
    cmap="Greens",
    annot=True, 
    fmt="d",
    linewidths=0.5,
    linecolor="white"
)

plt.xticks(ticks=np.arange(0.5, len(country_labels) + 0.5), labels=country_labels, rotation=0)

# Removed x and y labels as they are intuitive
plt.xlabel("")
plt.ylabel("")
plt.title("Top 10 Programming Languages in Countries")
plt.show()
```
