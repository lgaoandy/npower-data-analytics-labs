### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
# Examine both "LanguageHaveWorkedWith" and "LanguageWantToWorkWith"
print(df["LanguageHaveWorkedWith"].unique())
df["LanguageWantToWorkWith"].unique()
```

```python
# There are just too many languages in these lists, thus we target some specific ones:
languages_have = df["LanguageHaveWorkedWith"].dropna().str.split(";").explode()
languages_want = df["LanguageWantToWorkWith"].dropna().str.split(";").explode()

unique_languages = set(languages_have) | set(languages_want)
unique_languages
```

```python
# Target languages we want to compare
target_languages = [
    "JavaScript",
    "Python",
    "SQL",
    "TypeScript",
    "C#",
    "Go",
    "Perl",
    "Ruby",
    "R",
    "Rust",
    "Kotlin"
]

# Total number of respondents
total = len(df)

# Count how many respondents have each language
have_counts = {}
want_counts = {}

for language in target_languages:
    have_counts[language] = df["LanguageHaveWorkedWith"].fillna("").str.split(";").apply(lambda x: language in x).sum()
    want_counts[language] = df["LanguageWantToWorkWith"].fillna("").str.split(";").apply(lambda x: language in x).sum()

print(have_counts)
print(want_counts)

# Convert counts to percentage of the entire dataset
have_pct = {language: count / total * 100 for language, count in have_counts.items()}
want_pct = {language: count / total * 100 for language, count in want_counts.items()}

# Create dataframe for plotting
df_language_distribution = pd.DataFrame({
    "Language": target_languages,
    "Have": [have_pct[x] for x in target_languages],
    "Want": [want_pct[x] for x in target_languages]
})

df_language_distribution
```

```python
plot_data = df_language_distribution.set_index("Language")[["Have", "Want"]]
plot_data
```

```python
ax = plot_data.plot(
    kind="bar",
    figsize=(10, 5),
    color=["steelblue", "lightsteelblue"]
)

# Highlight Want bars where Want > Have
for i, (have, want) in enumerate(zip(plot_data["Have"], plot_data["Want"])):
    if want > have:
        ax.containers[1][i].set_color("orange")
        
for container in ax.containers:
    for bar in container:
        height = bar.get_height()
        if height < 12.5:
            ax.text(
                bar.get_x() + bar.get_width() / 2,
                height + 1,
                f"{height:.0f}%",
                ha="center",
                va="bottom",
                rotation=90,
                color="black"
            )
        else:
            ax.text(
                bar.get_x() + bar.get_width() / 2,
                height - 2,
                f"{height:.0f}%",
                ha="center",
                va="top",
                rotation=90,
                color="white"
            )

plt.xlabel("Programming Languages")
plt.ylabel("Percentage of Respondents (%)")
plt.title("Programming Language Usage vs. Desired Languages")
plt.xticks(rotation=0, ha="center")
plt.legend(title="Language Status")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
# First, investigate the countries with the most datasets:
df["Country"].value_counts().head(10)
```

```python
# Let's take the top 10 countries only
countries = [
    "United States of America",
    "Germany",
    "India",
    "United Kingdom of Great Britain and Northern Ireland",
    "Ukraine",
    "France",
    "Canada",
    "Poland",
    "Netherlands",
    "Brazil"
]

# Filter dataframe down to the 10 countries
df_top10 = df[df["Country"].isin(countries)]

# Cross country with remote work
df_top10_remote = pd.crosstab(
    df_top10["Country"],
    df_top10["RemoteWork"]
)

# Adjust remote work types (remove description on hybrid value)
df_top10_remote.columns = ["Hybrid", "In-Person", "Remote"]

# Adjust country names (United Kingdom and United States of America to UK and USA)
df_top10_remote = df_top10_remote.rename(index={
    "United Kingdom of Great Britain and Northern Ireland": "UK",
    "United States of America": "USA",
})

df_top10_remote
```

```python
plt.figure(figsize=(10, 5))

sns.heatmap(
    data=df_top10_remote,
    annot=True,
    fmt=",.0f",
    cmap="Blues",
    linewidths=0.5,
    linecolor="white"
)

plt.title("Remote Work Distribution by Country")
plt.xlabel("")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
# Distribution inside the prev heatmap shows it is heavily skewed to USA, due to its high volume of job postings
# This may not show the distribution of remote jobs in general but more so on "which country has more jobs"
# To fix this, we will calculate the percentage of remote work types within each country

df_top10_remote_pct = pd.crosstab(
    df_top10["Country"],
    df_top10["RemoteWork"],
    normalize="index"
) * 100

# Adjust remote work types (remove description on hybrid value)
df_top10_remote_pct.columns = ["Hybrid", "In-Person", "Remote"]

# Adjust country names (United Kingdom and United States of America to UK and USA)
df_top10_remote_pct = df_top10_remote_pct.rename(index={
    "United Kingdom of Great Britain and Northern Ireland": "UK",
    "United States of America": "USA",
})

df_top10_remote_pct
```

```python
# Sort by remote percentage
df_top10_remote_pct = df_top10_remote_pct.sort_values(
    by="Remote",
    ascending=False
)

plt.figure(figsize=(10, 5))

sns.heatmap(
    data=df_top10_remote_pct,
    annot=True,
    fmt=",.1f",
    cmap="Blues",
    linewidths=0.5,
    linecolor="white"
)

ax = plt.gca()
for text in ax.texts:
    text.set_text(text.get_text() + "%")

plt.title("Remote Work Distribution by Country")
plt.xlabel("")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
# Now from percentage, we can view how countries differ in remote work distribution. Ukraine has the highest remote work percentage
```
